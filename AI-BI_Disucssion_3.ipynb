### Loading Dataset and Preparing Features and Target 

```python
# =============================================================================
# AI_Discussion3.ipynb
# MGT 661 - Artificial Intelligence & Business Analytics
# Northwood University | DeVos Graduate School | Fall 2026
# Week 3 Discussion
# =============================================================================

# --- Imports ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from  sklearn.tree import DecisionTreeClassifier, export_text, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.datasets import load_iris
from sklearn import tree
# =============================================================================
# We will be using the Iris dataset provided by sklearn. This dataset looks at features of flowers to identify their species
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names) # We load the dataset as a pandas dataframe for use with the model and analysis 

df["species"] = iris.target # Labeling the target variable as species, 
print(df.head()) # Showing what the column names are; species shows a particular species of flower, 0 = Setosa, 1 = Versicolor, 3 = Virginica
# We can see from this output that there are 4 features, each being quantitative and being measured in centimeters (cm)
X = df.drop(columns=["species"]) # We identify the different columns of features to be used as the independent or X variables; we will be
# using these to predict the target or species of the flower.
y = df["species"] # Identifying the species as the dependent or y variable
# =============================================================================
```

### Train/Test Split

```python
# =============================================================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state =42, stratify=y) # Here we split the dataset into training and testing sets. The test size is set to 30% of
# the data leaving 70% for use in training, so we are using a 70/30 training/test split. The random_state=42 ensures that the data selected will be
#consistently random
print(f"\nFeatures used: {X.columns.tolist()}") # Showing specifically what features will be used in our model for prediction of species
print(f"\nTraining set size: {X_train.shape[0]} observations") # Showing the training set size
print(f"\nTesting set size: {X_test.shape[0]} observations") # Showing the testing set size
```

### Fitting a Decision Tree & Showing Results

```python
# =============================================================================
tree_model = DecisionTreeClassifier(max_depth=3, random_state=42) # For this classification problem we will be using a Decision Tree with a max depth
# of 3 initially
tree_model.fit(X_train, y_train) # Using our training data with the model

y_pred = tree_model.predict(X_test) # Generating the predictions for the test data set; this will be compared with the true values stored in y_test, 
# to measure the accuracy of the predictions

print(f"\nTraining Accuracy:   {accuracy_score(y_train, tree_model.predict(X_train)):.4f}") # Showing how well the model performed with training data
print(f"\nTest Accuracy:   {accuracy_score(y_test, y_pred):.4f}") # Showing how well the model perfomed on the test data

print("\n" + "=" * 60)

print("\nConfusion Matrix:") # A Confusion matrix is created to better explain how well the model performed beyond accuracy measures. This shows the type
# of predictions and errors made.
cm = confusion_matrix(y_test, y_pred)
cm_df = pd.DataFrame(cm, index=iris.target_names,
                     columns=iris.target_names)
print("  Rows = Actual | Columns = Predicted")
print(cm_df)
```

### Plotting the Decision Tree & Displaying Feature Importance 

```python
# =============================================================================
# We will plot the decision tree itself by first defining the feature names (fn) and class names (cn)
fn = ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']
cn = ['setosa', 'versicolor', 'virginica']
fig, axes = plt.subplots(nrows =1, ncols=1, figsize = (4,4), dpi = 300) # Adding clarity to the plot

tree.plot_tree(tree_model, feature_names=fn, class_names=cn, filled=True); # Plotting the decision tree with filled in colors for final predictions made
# Showing the feature importance that is used by the decision tree, this allows us to see what data the decision tree is using in its logic and how important
# each feature is for prediction.
importances = pd.Series(tree_model.feature_importances_, index=fn)
importances_sorted = importances.sort_values(ascending=False) # Sorted from most important to least important with ascending=False

print("\nFeature Importance Scores:")
print(importances_sorted.to_string())
print("\n" + "=" * 60)
# =============================================================================
```

### Testing the Effect of Tree Depth

```python
# =============================================================================
# Here we will be testing how changing the maximum depth allowed for the decision tree affects model accuracy, from 1 to 5.
max_depth_range = list(range(1,6))
# We will be showing training and test accuracy. A model may achieve very high training accuracy but lower test accuracy due
# to overfitting the training data. This means it is identifying patterns that are not reproducible on new, unseen data, such as test data.
train_accuracy = []  
test_accuracy = []

for depth in max_depth_range:
    tree_model_test = DecisionTreeClassifier(max_depth = depth,
                                        random_state=42) # Same process as training and fitting data, changed model name to reflect this is for max_depth testing
    tree_model_test.fit(X_train, y_train)

    train_accuracy.append(tree_model_test.score(X_train, y_train)) # Generating training accuracy
    test_accuracy.append(tree_model_test.score(X_test, y_test)) # Generating Testing accuracy
# Putting results in a pandas dataframe to be printed
results = pd.DataFrame({"Max Depth": max_depth_range,
                        "Train Accuracy": train_accuracy,
                        "Test Accuracy": test_accuracy})
print("\nMaximum Depth Setting Test:")
print(results.to_string(index=False))
print("\n" + "=" * 60)
```
